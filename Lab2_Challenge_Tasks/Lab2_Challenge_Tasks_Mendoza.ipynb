```python
# [EXERCISE 1] SENSOR READING VALIDATOR
import random
random.seed(SEED_NUM if 'SEED_NUM' in dir() else 0)

ENGINEER = "Mendoza"
SEED_NUM = 0

NUM_READINGS = SEED_NUM + 5
raw_readings = [round(random.uniform(0, 100), 1) for _ in range(NUM_READINGS)]
raw_readings.insert(2, "ERR")  # inject one invalid entry

print(f"Generated Sensor Data: {raw_readings}")

valid, invalid = [], []
for r in raw_readings:
    try:
        val = float(r)
        valid.append(val)
    except ValueError:
        invalid.append(r)

print(f"Valid Readings: {valid}")
print(f"Invalid Entries: {invalid}")

print("--- Classification ---")
for v in valid:
    if v < 30:
        cls = "LOW"
    elif v > 70:
        cls = "HIGH"
    else:
        cls = "NORMAL"
    print(f"Reading {v} -> {cls}")

print(f"Final Summary: Valid={len(valid)}, Invalid={len(invalid)}")
```

```python
# [EXERCISE 2] SIGNAL CHARACTER DIAGNOSTIC
signal = f"{ENGINEER}{SEED_NUM}"
print(f"Generated Signal: {signal}")

processed = signal.strip().upper()
print(f"Processed Signal: {processed}")

letters = digits = others = 0
print("--- Character Analysis ---")
for ch in processed:
    if ch.isalpha():
        kind = "LETTER"; letters += 1
    elif ch.isdigit():
        kind = "DIGIT"; digits += 1
    else:
        kind = "OTHER"; others += 1
    print(f"'{ch}' -> {kind}")

classification = "Alphanumeric Signal" if letters and digits else "Uniform Signal"
print(f"Signal Classification: {classification}")
print(f"Final Summary: Letters={letters}, Digits={digits}, Others={others}")
```

```python
# [EXERCISE 3] ACCESS CONTROL WITH ATTEMPT LIMIT
PASSWORD = f"{ENGINEER.upper()}{SEED_NUM}"
ATTEMPT_LIMIT = SEED_NUM + 3
attempts = 0
result = "SYSTEM LOCKED"

print(f"Generated Password: {PASSWORD}")
print(f"Attempt Limit: {ATTEMPT_LIMIT}")

while attempts < ATTEMPT_LIMIT:
    pwd = input("Enter password: ")
    attempts += 1
    if pwd == PASSWORD:
        result = "ACCESS GRANTED"
        print("ACCESS GRANTED")
        break
    else:
        print(f"ACCESS DENIED ({attempts}/{ATTEMPT_LIMIT})")

if result == "SYSTEM LOCKED":
    print("SYSTEM LOCKED: Too many failed attempts.")

print(f"Final Attempts Used: {attempts}")
print(f"Final Result: {result}")
```
